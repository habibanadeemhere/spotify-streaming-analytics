```python
# SPOTIFY MUSIC TRENDS - EXPLORATORY DATA ANALYSIS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load final cleaned dataset
df = pd.read_csv("data/spotify_final.csv")

# Set graph style
sns.set_theme(style="whitegrid")

print("SPOTIFY MUSIC TRENDS - EDA PROJECT")
print("=" * 50)

print("Dataset Shape:", df.shape)
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nFirst 5 Rows:")
display(df.head())
```


# 1. Understanding the Dataset

## Dataset Source
The dataset is called **Most Streamed Spotify Songs 2024**. It was originally published on Kaggle.

Dataset link: https://www.kaggle.com/datasets/nelgiriyewithana/most-streamed-spotify-songs-2024

## About the Dataset
This dataset contains information about popular songs, artists, streaming numbers, playlist counts, and music popularity across different platforms.

Each row represents a song entry in the dataset.

## Dataset Dimensions
- Total rows: 4,593
- Total columns: 28
- Data type: Numerical and categorical


```python

# Display all column names
print("COLUMN NAMES")
for col in df.columns:
    print(col)

# Display data types
print("\nDATA TYPES")
print(df.dtypes)

# Dataset information
print("\nDATASET INFORMATION")
df.info()

# Statistical summary
print("\nSTATISTICAL SUMMARY")
display(df.describe())
```

```python

# Check missing values in final dataset

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
display(missing_values[missing_values > 0])

print("Total Missing Values:", df.isnull().sum().sum())

# Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

```


# 2. Univariate Analysis

Univariate analysis means studying one variable at a time.

In this section, we will examine:
- Spotify Popularity distribution.
- Track Score distribution.
- Distribution of Spotify Streams.
- Explicit and non-explicit song counts.


```python

# Missing values before cleaning

original_df = pd.read_csv(
    "data/spotify.csv",
    encoding="latin1",
    low_memory=False
)

missing = original_df.isnull().sum()
missing = missing[missing > 0].sort_values()

plt.figure(figsize=(12, 7))
missing.plot(kind="barh")

plt.title("Missing Values Before Data Cleaning")
plt.xlabel("Number of Missing Values")
plt.ylabel("Columns")
plt.tight_layout()
plt.show()
```

```python

# Outlier detection using boxplots

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df["Track Score"])
plt.title("Outliers in Track Score")

plt.subplot(1, 2, 2)
sns.boxplot(y=df["Spotify Streams"])
plt.title("Outliers in Spotify Streams")

plt.tight_layout()
plt.show()
```


# 3. Exploratory Data Analysis – Eight Questions

We will investigate Spotify streaming trends, artist popularity, social media engagement, and playlist presence using eight analytical questions.


```python

# QUESTION 1: Which songs have the highest Spotify streams?

top_songs = df.nlargest(10, "Spotify Streams")

display(top_songs[["Track", "Artist", "Spotify Streams"]])

plt.figure(figsize=(12, 6))
sns.barplot(data=top_songs, x="Spotify Streams", y="Track")
plt.title("Top 10 Songs by Spotify Streams")
plt.tight_layout()
plt.show()


# QUESTION 2: Which artists have the most songs?

top_artists = df["Artist"].value_counts().head(10)

display(top_artists)

top_artists.plot(kind="bar", figsize=(10, 5))
plt.title("Top 10 Artists by Number of Songs")
plt.xlabel("Artist")
plt.ylabel("Number of Songs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# QUESTION 3: What is the distribution of Spotify popularity?

print(df["Spotify Popularity"].describe())

plt.figure(figsize=(10, 5))
sns.histplot(df["Spotify Popularity"], bins=20, kde=True)
plt.title("Spotify Popularity Distribution")
plt.show()


# QUESTION 4: Is Track Score related to Spotify Streams?

print("Correlation:", df["Track Score"].corr(df["Spotify Streams"]))

plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x="Track Score", y="Spotify Streams")
plt.title("Track Score vs Spotify Streams")
plt.show()


# QUESTION 5: Are YouTube Views related to Spotify Streams?

print("Correlation:", df["YouTube Views"].corr(df["Spotify Streams"]))

plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x="YouTube Views", y="Spotify Streams")
plt.title("YouTube Views vs Spotify Streams")
plt.show()


# QUESTION 6: Which songs have the highest TikTok views?

top_tiktok = df.nlargest(10, "TikTok Views")

display(top_tiktok[["Track", "Artist", "TikTok Views"]])

plt.figure(figsize=(12, 6))
sns.barplot(data=top_tiktok, x="TikTok Views", y="Track")
plt.title("Top 10 Songs by TikTok Views")
plt.tight_layout()
plt.show()


# QUESTION 7: Does explicit content relate to Spotify popularity?

explicit_data = df.groupby("Explicit Track")["Spotify Popularity"].mean()

explicit_data.index = ["Non-Explicit", "Explicit"]

display(explicit_data)

explicit_data.plot(kind="bar", figsize=(8, 5))
plt.title("Average Popularity: Explicit vs Non-Explicit")
plt.ylabel("Average Spotify Popularity")
plt.xticks(rotation=0)
plt.show()


# QUESTION 8: Which platform has the highest playlist count?

platforms = [
    "Spotify Playlist Count",
    "Apple Music Playlist Count",
    "Deezer Playlist Count",
    "Amazon Playlist Count"
]

platform_means = df[platforms].mean()

display(platform_means)

platform_means.plot(kind="bar", figsize=(10, 5))
plt.title("Average Playlist Count Across Platforms")
plt.ylabel("Average Playlist Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```


# 4. Answers and Key Findings

1. **Top Spotify Songs:** Blinding Lights by The Weeknd has the highest Spotify streams in this dataset, with approximately 4.28 billion streams.

2. **Top Artists:** Taylor Swift and Drake have the highest number of song entries, with 63 each.

3. **Spotify Popularity:** The average Spotify popularity score is approximately 64.11.

4. **Track Score and Streams:** The correlation is 0.247, showing a weak positive relationship.

5. **YouTube and Spotify:** The correlation is 0.426, showing a moderate positive relationship between YouTube views and Spotify streams.

6. **TikTok Views:** Monkeys Spinning Monkeys by Kevin MacLeod has the highest TikTok views in the dataset.

7. **Explicit Content:** Explicit songs have an average popularity of 65.39, compared with 63.39 for non-explicit songs. This shows a difference, not proof that explicit content causes popularity.

8. **Playlist Presence:** Spotify has the highest average playlist count among the four platforms compared.
